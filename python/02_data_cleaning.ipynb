# 1. Load Libraries and Dataset

```python
import pandas as pd 
import numpy as np 

df = pd.read_csv('../data/raw/telco-Customer-Churn.csv')

```

# 2. Inspect Data

```python
df.columns
```

# 3. Converting to the correct data type 

`TotalCharges` is stored as a string and requires conversion to a numeric data type.

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'] , errors= 'coerce')
```

# 4. Check Missing Values in TotalCharges 

```python
df['TotalCharges'].isnull().sum()
```

```python
df[df['TotalCharges'].isnull() == True][['MonthlyCharges' , 'tenure','TotalCharges']]
```

We notice that the reason why the TotalCharges is null because the tenure is 0 which make sence . SO we are going to replace null with 0 

```python
df['TotalCharges'] = df['TotalCharges'].fillna(0)

df['TotalCharges'].isnull().sum()
```

# 5. Cleaning Churn 

Churn is stored as a string (`Yes` / `No`) and should be converted to a boolean field.

```python
df['Churn'] = df['Churn'].map({'Yes' : True , 'No' : False }).astype(bool)

df.dtypes
```

# 6. Converted SeniorCitizen to boolean

```python
df['SeniorCitizen'] = df['SeniorCitizen'].astype(bool)

df['SeniorCitizen'].value_counts()
```

# 7. Convert categorical fields to boolean 

Several categorical fields representing binary values (`Yes` / `No`) should be converted to boolean data types.

```python
cols = [
    'Partner',
    'Dependents',
    'PhoneService',
    'MultipleLines',
    'DeviceProtection',
    'OnlineSecurity',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies',
    'PaperlessBilling'
]


bol_col = []

for col in cols:
    print(f'{col}: {df[col].nunique()} unique values')
    print(df[col].unique())
    print('-' * 40)
    if df[col].nunique() == 2 :
        bol_col.append(col)


bol_col
```

```python
df[bol_col] = df[bol_col].replace({'Yes' : True , 'No': False}).astype(bool)

df[bol_col].dtypes
```

# 8. Validated final data types

```python
df.dtypes
```

# 9. Output 

```python
df.to_csv('../data/processed/telecom_churn_clean.csv')
```
